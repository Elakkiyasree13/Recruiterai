```python
import pandas as pd
import numpy as np

marks = [50,80,65,88,70,95,40,98,85,90]
df = pd.DataFrame(marks, columns=["Marks"])
df
```

```python
freq = pd.cut(df["Marks"], bins=5).value_counts().sort_index()
cum_freq = freq.cumsum()

print("Frequency:\n", freq)
print("\nCumulative Frequency:\n", cum_freq)
```

```python
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Mode:", df["Marks"].mode()[0])
```

```python
print("Range:", df["Marks"].max() - df["Marks"].min())
print("Standard Deviation:", df["Marks"].std())
```

```python
mean = df["Marks"].mean()
std = df["Marks"].std()

z_scores = (df["Marks"] - mean) / std
df["Z-Score"] = z_scores

df
```
